```python
df = pd.read_csv("ExamStataData.csv")   # adjust extension/sheet_name if needed
df.head()
```

```python
# UK per-population variables + logs ---
df["uk_hpop"] = df["h_uk"] / df["pop15plus_uk"]
df["uk_kpop"] = df["k_uk"] / df["pop15plus_uk"]
df["uk_ypop"] = df["y_uk"] / df["pop15plus_uk"]

df["log_uk_hpop"] = np.log(df["uk_hpop"])
df["log_uk_kpop"] = np.log(df["uk_kpop"])
df["log_uk_ypop"] = np.log(df["uk_ypop"])

# US per-population variables + logs ---
df["us_hpop"] = df["h_us"] / df["pop15plus_us"]
df["us_kpop"] = df["k_us"] / df["pop15plus_us"]
df["us_ypop"] = df["y_us"] / df["pop15plus_us"]

df["log_us_hpop"] = np.log(df["us_hpop"])
df["log_us_kpop"] = np.log(df["us_kpop"])
df["log_us_ypop"] = np.log(df["us_ypop"])
```

```python
# --- Summarize labor shares (Stata: sum labsh_uk; sum labsh_us) ---
labsh_uk_desc = df["labsh_uk"].describe()
labsh_us_desc = df["labsh_us"].describe()

labsh_uk_mean = float(df["labsh_uk"].mean())
labsh_us_mean = float(df["labsh_us"].mean())

labsh_uk_desc, labsh_us_desc, labsh_uk_mean, labsh_us_mean
```

```python
# --- Constraints (Stata constraint statements) ---
# In Python, we'll just store these values and use them when running the constrained regressions.

alpha_uk = labsh_uk_mean  # mean UK labor share
alpha_us = labsh_us_mean  # mean US labor share

# Labor-side constraints:
#   coef on log_us_hpop is alpha_uk
#   coef on log_uk_hpop is alpha_us
b_us_h = alpha_uk
b_uk_h = alpha_us

# Capital-side constraints:
#   coef on log_us_kpop is 1 - alpha_uk
#   coef on log_uk_kpop is 1 - alpha_us
b_us_k = 1 - alpha_uk
b_uk_k = 1 - alpha_us

alpha_uk, alpha_us, b_us_h, b_uk_h, b_us_k, b_uk_k
```

```python
# --- Govt spending / GDP ratios (Stata: gen gratio_uk = g_uk/y_uk; etc.) ---
df["gratio_uk"] = df["g_uk"] / df["y_uk"]
df["gratio_us"] = df["g_us"] / df["y_us"]

# Stata: sum gratio*
df[[c for c in df.columns if c.startswith("gratio_")]].describe()
```

```python
import numpy as np
import statsmodels.formula.api as smf


for country in ['us', 'uk']:
   
    alpha_constrained = df[f'labsh_{country}'].mean()
    print(f"{country.upper()} alpha constraint: {alpha_constrained}")

    df[f'log_y_{country}'] = np.log(df[f'y_{country}'])
    df[f'log_h_{country}'] = np.log(df[f'h_{country}'])
    df[f'log_k_{country}'] = np.log(df[f'k_{country}'])
    
    # ln(Y) - alpha*ln(H) - (1-alpha)*ln(K) = ln(Z_bar) + ln(z_t) 
    df[f'dep_var_{country}'] = (df[f'log_y_{country}'] 
                                - alpha_constrained * df[f'log_h_{country}'] 
                                - (1 - alpha_constrained) * df[f'log_k_{country}'])

    results = smf.ols(formula=f'dep_var_{country} ~ 1', data=df).fit()
    

    ln_Z_bar = results.params['Intercept']
    Z_bar_estimate = np.exp(ln_Z_bar)
    p_value = results.pvalues['Intercept']
    
    print(f"{country.upper()} ln(Z_bar) Estimate: {ln_Z_bar}")
    print(f"P-value: {p_value}")

   
    df[f'log_{country}_zpop'] = results.resid
```
